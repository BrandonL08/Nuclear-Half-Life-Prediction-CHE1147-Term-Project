```python
#Load and inspect the ENSDF-derived dataset

import pandas as pd
import numpy as np

df = pd.read_csv("ensdf_ground_state_candidates.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# How many rows were extracted?
print("Total candidate rows:", len(df))


# Check whether any nuclides appear more than once.
print("Duplicate nuclides:", df.duplicated(subset=["nuclide"]).sum())


# Check the identity relationship A = Z + N.
identity_errors = df[df["A"] != df["Z"] + df["N"]]

print("Identity errors:", len(identity_errors))


# Find ambiguous ground states.
ambiguous = df[df["ground_state_ambiguous"] == True]

print("Ambiguous ground states:", len(ambiguous))


# Check half-life coverage.
from collections import Counter

half_life_counts = Counter(df["half_life_status"])

print("Half-life status counts:", half_life_counts)
```

```python
# Check feature coverage.

features = [
    "A",
    "Z",
    "half_life_seconds",
    "spin_parity_raw",
    "Q_alpha_keV",
    "Q_beta_keV",
    "S_n_keV",
    "S_p_keV",
]

print("Feature coverage")
print("-" * 45)

for feature in features:
    available = sum(
        df[feature].notnull() & (df[feature] != "")
    )

    total = len(df)
    percentage = 100 * available / total if total else 0

    print(
        f"{feature:25s} "
        f"{available:6d}/{total:<6d} "
        f"({percentage:.1f}%)"
    )

# Create a bar chart to visualize the coverage of each feature
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
for feature in features:
    available = sum(
        df[feature].notnull() & (df[feature] != "")
    )

    total = len(df)
    percentage = 100 * available / total if total else 0

    plt.bar(feature, percentage)
plt.xlabel("Features")
plt.ylabel("Coverage (%)")
plt.title("Feature Coverage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print(
    df["half_life_status"]
    .value_counts(dropna=False)
)

# Make pie chart of half-life status distribution with percentages included with labels 
import matplotlib.pyplot as plt

df["half_life_status"].value_counts(dropna=False).plot(kind="pie", labels=df["half_life_status"].value_counts(dropna=False).index + " (" + df["half_life_status"].value_counts(dropna=False).apply(lambda x: f"{x/len(df)*100:.1f}%") + ")")
plt.title("Half-Life Status Distribution")
plt.ylabel("")
plt.show()
```

```python
#Create a summary of the Q, S and half-life values to understand their distributions
print(df[["Q_beta_keV", "Q_alpha_keV", "S_n_keV", "S_p_keV", "half_life_seconds"]].describe())

# Create a boxplot for each of the Q, S and half-life values to visualize their distributions  
plt.figure(figsize=(8, 6))
df.boxplot(column=["Q_beta_keV", "Q_alpha_keV", "S_n_keV", "S_p_keV", "half_life_seconds"])
plt.title("Boxplots of Nuclear Data")
plt.ylabel("Value")
plt.show()
```

```python
# Plot histograms for Q values to visualize their distributions.

features = [
    "Q_alpha_keV",
    "Q_beta_keV"
]

for feature in features:
    available = sum(
        df[feature].notnull() & (df[feature] != "")
    )

    total = len(df)
    percentage = 100 * available / total if total else 0

    print(
        f"{feature:25s} "
        f"{available:6d}/{total:<6d} "
        f"({percentage:.1f}%)"
    )

# Create a bar chart to visualize the coverage of each feature
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
for feature in features:
    available = sum(
        df[feature].notnull() & (df[feature] != "")
    )

    total = len(df)
    percentage = 100 * available / total if total else 0

    plt.bar(feature, percentage)
plt.xlabel("Features")
plt.ylabel("Coverage (%)")
plt.title("Feature Coverage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 8))
df[["Q_beta_keV", "Q_alpha_keV"]].hist(bins=30, figsize=(12, 8))
plt.show()
```

```python
# Plot neutron-to-proton ratio (N/Z) vs. half-life (in seconds) to visualize the relationship between mass number and half-life
plt.figure(figsize=(10, 6))
plt.scatter(df[df["Z"] > 0]["N"] / df[df["Z"] > 0]["Z"], (df[df["Z"] > 0]["half_life_seconds"]), alpha=0.5)
plt.xlabel("Neutron-to-Proton Ratio (N/Z)")
plt.ylabel("Half-Life (seconds)")
plt.yscale("log")
plt.title("Relationship between Neutron-to-Proton Ratio and Half-Life")
plt.show()
```

```python
# Plot mass number (A) vs. half-life (in seconds) to visualize the relationship between mass number and half-life
plt.figure(figsize=(10, 6))
plt.scatter(df[df["Z"] > 0]["A"], (df[df["Z"] > 0]["half_life_seconds"]), alpha=0.5)
plt.xlabel("Mass Number (A)")
plt.ylabel("Half-Life (seconds)")
plt.yscale("log")
plt.title("Relationship between Mass Number and Half-Life")
plt.show()
```

```python
# Plot Q_beta_kev vs. half-life (in seconds) to visualize the relationship between Q_beta and half-life
plt.figure(figsize=(10, 6))
plt.scatter(df[df["Q_beta_keV"].notnull()]["Q_beta_keV"], (df[df["Q_beta_keV"].notnull()]["half_life_seconds"]), alpha=0.5)
plt.xlabel("Q_beta_keV")
plt.ylabel("Half-Life (seconds)")
plt.yscale("log")
plt.title("Relationship between Q_beta_keV and Half-Life")
plt.show()
```
