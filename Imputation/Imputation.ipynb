```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

eda_df = pd.read_csv("eda_df.csv")
imputed_df = eda_df.copy()

energy_features = [
    "A",
    "N/Z",
    "Q_beta_keV",
    "Q_alpha_keV",
    "S_n_keV",
    "S_p_keV"
]

imputer = SimpleImputer(strategy="median")

imputed_df[energy_features] = imputer.fit_transform(
    imputed_df[energy_features]
)

imputed_df.head()
imputed_df.to_csv("imputed_df.csv", index=False)
```
