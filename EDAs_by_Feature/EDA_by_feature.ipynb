```python
import pandas as pd
import numpy as np

eda_df = pd.read_csv("https://raw.githubusercontent.com/BrandonL08/Nuclear-Half-Life-Prediction-CHE1147-Term-Project/refs/heads/main/eda_df.csv")

print("Dataset shape:", eda_df.shape)
display(eda_df.head())
```

```python
# Isotopes with a recorded Q_beta value
df_with_QB = eda_df[eda_df["Q_beta_keV"].notna()]

# Isotopes without a recorded Q_beta value
df_without_QB = eda_df[eda_df["Q_beta_keV"].isna()]

print("With Q_beta:", len(df_with_QB))
print("Without Q_beta:", len(df_without_QB))

columns = ["Z", "N", "log_half_life"]

print("WITH Q_beta")
print(df_with_QB[columns].describe().round(2))

print("\nWITHOUT Q_beta")
print(df_without_QB[columns].describe().round(2))


import matplotlib.pyplot as plt

columns = ["Z", "N", "log_half_life"]

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for ax, col in zip(axes, columns):
    ax.boxplot([
        df_with_QB[col].dropna(),
        df_without_QB[col].dropna()
    ], tick_labels=["With Q_beta", "Without Q_beta"])

    ax.set_title(col)
    ax.set_ylabel("Value")

plt.tight_layout()
plt.show()



```

```python
# Isotopes with a recorded S_n_keV value
df_with_Sn = eda_df[eda_df["S_n_keV"].notna()]

# Isotopes without a recorded S_n_keV value
df_without_Sn = eda_df[eda_df["S_n_keV"].isna()]

print("With S_n:", len(df_with_Sn))
print("Without S_n:", len(df_without_Sn))

columns = ["Z", "N", "log_half_life"]

print("WITH S_n")
print(df_with_Sn[columns].describe().round(2))

print("\nWITHOUT S_n")
print(df_without_Sn[columns].describe().round(2))


import matplotlib.pyplot as plt

columns = ["Z", "N", "log_half_life"]

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for ax, col in zip(axes, columns):
    ax.boxplot([
        df_with_Sn[col].dropna(),
        df_without_Sn[col].dropna()
    ], tick_labels=["With S_n", "Without S_n"])

    ax.set_title(col)
    ax.set_ylabel("Value")

plt.tight_layout()
plt.show()
```
